```python
#1. List Creation

#Create a list named age_list with five integer elements
age_list=[24,25,27,28,29,30]
print(age_list)

#Create a list named name_list with five string elements
name_list=["Malathy","Gopi","Divya","Nage","Ani"]
print(name_list)
```

```python
#List Operations / Modifications

#Append the string "Yazhini" to name_list
name_list.append("yazhini")
print(name_list)

#Insert the element 26 at index 2 in age_list.
age_list.insert(2,26)
print(age_list)

#Remove the string "Yazhini" from name_list.
name_list.remove("yazhini")
print(name_list)

#Pop the last element from age_list.
age_list.pop()
print(age_list)

#Extend the age_list with additional ages [31,30,32].
age_list.extend([31,30,32])
print(age_list)

#Sort age_list in descending order.
age_list.sort(reverse=True)

#Find the max age, min age, and sum of all ages from age_list.
max(age_list)
min(age_list)
sum(age_list)
```

```python
#Accessing List Elements:

#Print the first element of name_list.
print(name_list[0])

#Print the last element of name_list.
print(name_list[-1])

#Print the elements from index 2 to index 4 in name_list.
print(name_list[2:5])

#Print the elements of name_list in reverse order.
print(name_list[::-1])
```

```python
#Dictionary (Creation, Modification and Access):

#Create a dictionary named student_marks that maps the names of five students to their marks (use a scale of 0 to 100).
students_mark={"nivi":99,"kavi":55,"pavi":78,"thiru":66,"navi":83}
print(students_mark["navi"])

#Add a new student, "Janani," with a mark of 80 to the student_marks dictionary.
students_mark["janani"]=80
print(students_mark["janani"])

#Update the mark of any one older student to 82.
students_mark["kavi"]=82
print(students_mark["kavi"])

#Use the keys(), values(), and items() methods to print all keys, values, and key-valuepairs in the student_marks dictionary.
print(students_mark.keys())
print(students_mark.values())
print(students_mark.items())

```
