**1.Creating Pandas Series**

```python
marks=pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
marks
```

**2.Indexing and Slicing**

2a.Use the integer index position to access the mark of the 1st rank student.

```python
marks.iloc[0]
```

2b.Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.

```python
marks.loc[["Rank1","Rank2","Rank3"]]
```

2c.Use the iloc accessor to retrieve the mark of the 3rd-ranked student

```python
marks.iloc[2]
```

2d.Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.

```python
marks[marks>90]
```

**3.Manipulating Series**

3a.Modify the mark of the student with the 1st rank to 100.

```python
marks.loc["Rank1"]=100
marks
```

3b.Remove the entry corresponding to the last rank from the Series.

```python
marks=marks.drop("Rank5")
marks
```

3c.Compute the CGPA by dividing each mark in the Series by 10.

```python
cgpa=marks/10
cgpa
```
